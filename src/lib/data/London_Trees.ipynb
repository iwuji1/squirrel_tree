```python
tree_Age_p10 = tree_borough_join["girth"].quantile(0.1)
tree_Age_p90 = tree_borough_join["girth"].quantile(0.9)
```

```python
def p10_p90(series: pd.Series) -> pd.Series:
    s=series[series > 0]
    if len(s) < 5:
        return pd.Series({"p10": np.nan, "p90": np.nan})
    return pd.Series({"p10": s.quantile(0.10), "p90": s.quantile(0.90)})
```

```python
borough_stats = tree_borough_join.groupby("borough")["girth"].apply(p10_p90).reset_index()
```

```python
borough_stats = borough_stats.pivot(index="borough", columns="level_1", values="girth").reset_index()
```

```python
borough_stats.rename(columns={"p10": "P10_girth_borough", "p90": "P90_girth_borough"}, inplace=True)
```

```python
tree_borough_join_G = pd.merge(tree_borough_join, borough_stats, on="borough", how='left')
```

```python
#Girth & AgeFactor
denom = (tree_borough_join_G["P90_girth_borough"] - tree_borough_join_G["P10_girth_borough"]).replace(0, np.nan)
raw = (tree_borough_join_G["girth"] - tree_borough_join_G["P10_girth_borough"]) / denom
tree_borough_join_G["GirthIndex"] = raw.clip(lower=0, upper=1).fillna(0)

tree_borough_join_G["AgeFactor"] = 0.9 + 0.4 * tree_borough_join_G["GirthIndex"]
```

```python
tree_borough_join_G.head()
```

```python
tree_borough_join_G["TreeRent"] = tree_borough_join_G["Borough_Nut_Rent"]*tree_borough_join_G["AgeFactor"]
```

```python
tree_borough_join_G.head()
```

```python
tree_borough_join_G["CanopyScore"] = tree_borough_join_G["canopy"] / max(tree_borough_join_G["canopy"])
tree_borough_join_G["HeightScore"] = tree_borough_join_G["height"] / max(tree_borough_join_G["height"])
tree_borough_join_G["GirthScore"] = tree_borough_join_G["girth"] / max(tree_borough_join_G["girth"])
```

```python
tree_borough_join_G["StructureScore"] = (0.4 * tree_borough_join_G["CanopyScore"]) + (0.3*tree_borough_join_G["HeightScore"]) + (0.3*tree_borough_join_G["GirthScore"])
```

```python
tree_borough_join_G["FoodScore"] = tree_borough_join_G["Food squirrels can eat (Yes / No)"].str.strip().str.lower().map({
    "yes": 1,
    "no": 0
}).fillna(0).astype(int)
```

```python
tree_borough_join_final = pd.merge(tree_borough_join_G, df_treeLoc, on="location", how='left')
```

```python
tree_borough_join_final.head()
```

```python
tree_borough_join_final["Tree Value Score"] = 100*((0.5*tree_borough_join_final["StructureScore"])+(0.3*tree_borough_join_final["FoodScore"])+(0.2*tree_borough_join_final["SiteFactor"]))
```

```python
import pandas as pd
import json

# --- 0) Pick the base dataframe ONCE (so you don't mix df vs tree_borough_join_final)
base = tree_borough_join_final

# --- 1) Top 10 trees per borough
top10 = (
    base.sort_values(["borough", "Tree Value Score"], ascending=[True, False])
        .groupby("borough", as_index=False)
        .head(10)
        .reset_index(drop=True)
)

# Choose fields (REMOVE duplicates like "location" twice)
tree_cols = [
    "borough", "uniqueid", "Tree", "taxon_species", "lat", "lon",
    "TreeRent", "location", "height", "girth",
    "Food squirrels can eat (Yes / No)", "Tree Value Score", "Rank"
]

# Keep only existing cols
tree_cols = [c for c in tree_cols if c in top10.columns]

# IMPORTANT: ensure the dataframe has unique column names (drop duplicate columns)
top10_trees = top10.loc[:, ~top10.columns.duplicated()].copy()

# Now select your fields (also ensure uniqueness here too)
top10_trees = top10_trees[tree_cols].copy()

# --- 2) Borough summary layer (use SAME base df)
borough_summary = (
    base.groupby("borough")
        .agg(
            tree_count=("uniqueid", "count"),
            most_common_tree=("Tree", lambda s: s.mode().iat[0] if not s.mode().empty else None)
        )
        .reset_index()
)

# --- 3) Nest trees WITHOUT groupby.apply (avoids deprecation warning)
trees_by_borough = {
    b: g.drop(columns=["borough"], errors="ignore").to_dict(orient="records")
    for b, g in top10_trees.groupby("borough")
}

# --- 4) Build final structure
output = {
    "boroughs": [
        {
            "borough": row["borough"],
            "tree_count": int(row["tree_count"]),
            "most_common_tree": row["most_common_tree"],
            "trees": trees_by_borough.get(row["borough"], [])
        }
        for _, row in borough_summary.iterrows()
    ]
}

# --- 5) Export JSON
with open("boroughs_with_top_trees.json", "w") as f:
    json.dump(output, f, indent=2)

print("Exported: boroughs_with_top_trees.json")
```
